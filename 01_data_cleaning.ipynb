```python
import re
import pandas as pd

def parse_cpu(name: str):
    """Extract brand, family, series, model, suffix from CPU string."""

    if pd.isna(name):
        return {
            "processor_brand": None,
            "processor_family": None,
            "processor_series": None,
            "processor_model": None,
            "processor_suffix": None,
        }

    s = str(name)
    s = re.sub(r"\s+", " ", s).strip()

    brand = family = series = model = suffix = None

    # ---------- BRAND ----------
    if re.search(r"\bIntel\b", s, re.IGNORECASE):
        brand = "Intel"
    elif re.search(r"\bAMD\b", s, re.IGNORECASE):
        brand = "AMD"
    elif re.search(r"\bApple\b", s, re.IGNORECASE):
        brand = "Apple"
    elif re.search(r"\bSnapdragon\b|\bQualcomm\b", s, re.IGNORECASE):
        brand = "Snapdragon"

    # ---------- FAMILY (brand-specific) ----------
    if brand == "Intel":
        fam_patterns = [
            (r"Core Ultra", "Core Ultra"),
            (r"\bCore\b", "Core"),
            (r"Pentium", "Pentium"),
            (r"Celeron", "Celeron"),
            (r"\bAtom\b", "Atom"),
            (r"Xeon", "Xeon"),
            # NEW: Intel N-series (N100, N200, etc.)
            (r"\bN\d{2,3}\b", "N-Series"),
        ]
    elif brand == "AMD":
        fam_patterns = [
            (r"Ryzen", "Ryzen"),
            (r"Athlon", "Athlon"),
            (r"EPYC", "EPYC"),
            (r"Threadripper", "Threadripper"),
            (r"A[\s-]?Series", "A-Series"),
            (r"\bFX\b", "FX"),
        ]
    elif brand == "Apple":
        fam_patterns = [
            (r"\bM[1-9]\b", "M-Series"),
        ]
    elif brand == "Snapdragon":
        fam_patterns = [
            (r"Snapdragon", "Snapdragon"),
        ]
    else:
        fam_patterns = []

    for pat, fam_val in fam_patterns:
        if re.search(pat, s, re.IGNORECASE):
            family = fam_val
            break

    # ---------- Series (brand-specific) ----------
    if brand == "Intel":
        m_series = re.search(
            r"(Core Ultra\s+[3579]|Core\s+i[3579]|Core\s+[3579]|Core\s+5|Core\s+7|Core\s+9|"
            r"Pentium|Celeron|Atom|Xeon)",
            s,
            re.IGNORECASE,
        )
        if m_series:
            series = m_series.group(0).title()

    elif brand == "AMD":
        m_ai = re.search(r"(Ryzen\s+AI\s+(?:3|5|7|9|MAX))", s, re.IGNORECASE)
        if m_ai:
            series = m_ai.group(0).title()
        else:
            m_series = re.search(r"(Ryzen\s+\d|Ryzen\s+[3-9])", s, re.IGNORECASE)
            if m_series:
                series = m_series.group(0).title()

    elif brand == "Apple":
        m_gen = re.search(r"\bM([1-9])\b", s, re.IGNORECASE)
        if m_gen:
            gen = m_gen.group(1)
            series = f"M{gen}"
            model = series

        m_model = re.search(r"\b(M[1-9])\s+(Pro|Max|Ultra)\b", s, re.IGNORECASE)
        if m_model:
            base = m_model.group(1).upper()
            suf = m_model.group(2).title()
            series = base
            model = f"{base} {suf}"
            suffix = suf

    elif brand == "Snapdragon":
        m_series = re.search(
            r"(Snapdragon\s+X|Snapdragon\s+8cx|Snapdragon\s+7c|Snapdragon)",
            s,
            re.IGNORECASE,
        )
        if m_series:
            series = m_series.group(0).title()

        after = re.split(r"Snapdragon(?:\s+X|\s+8cx|\s+7c)?", s, flags=re.IGNORECASE)
        if len(after) > 1:
            rest = after[1].strip(" -")
            if rest:
                model = rest

    # ---------- GENERIC MODEL DETECTION ----------
    if model is None:
        tokens = re.split(r"\s+", s)
        candidates = []
        for t in tokens:
            t_clean = t.strip(",()")
            if not re.search(r"\d", t_clean):
                continue
            if re.match(r"^\d+(st|nd|rd|th)$", t_clean.lower()):
                continue
            if re.match(r"^i[3579]$", t_clean.lower()):
                continue
            candidates.append(t_clean)

        if candidates:
            model = candidates[-1].upper()

    # ---------- Fill missing series from model ----------
    if model and series is None:
        if family:
            series = f"{family} {model}"
        else:
            series = model

    # ---------- Suffix from model ----------
    if model and suffix is None:
        m_suf = re.search(r"(\d+)([A-Za-z]+)$", model)
        if m_suf:
            suffix = m_suf.group(2).upper()

    return {
        "processor_brand": brand,
        "processor_family": family,
        "processor_series": series,
        "processor_model": model,
        "processor_suffix": suffix,
    }
```

```python
core_word_map = {
    "single": 1,
    "dual": 2,
    "triple": 3,
    "quad": 4,
    "penta": 5,
    "hexa": 6,
    "octa": 8,
    "deca": 10,
}

def extract_core_count(x):
    if pd.isna(x):
        return np.nan
    
    s = str(x).strip().lower()

    # 🚫 Ignore strings that look like CPU *names* not core configs
    # e.g. "13th Gen Intel Core i3", "Intel Core Ultra 5 228V"
    if ("intel core" in s or "core ultra" in s) and "cores" not in s:
        return np.nan

    # 1️⃣ Numeric pattern like "20 Cores", "10 Cores", "5 Cores (1P + 4E)"
    m_num = re.search(r"(\d+)\s+cores\b", s)
    if m_num:
        return int(m_num.group(1))

    # 2️⃣ Word-based pattern: "dual core", "quad core", "hexa core", "octa core"
    for word, value in core_word_map.items():
        # require "<word> core", not just the word alone
        if re.search(rf"\b{word}\s+core\b", s):
            return value

    # If nothing matches, it's probably not a real core config
    return np.nan
```

```python
df1["core_count"] = df1["cores"].apply(extract_core_count)
```

```python
def extract_threads(x):
    if pd.isna(x):
        return np.nan

    s = str(x).lower().strip()

    # explicit thread count e.g., "20 Threads", "12 threads"
    m = re.search(r'(\d+)\s*threads?', s)
    if m:
        return int(m.group(1))

    # If no thread info present → NaN
    return np.nan

df1["threads_count"] = df1["cores"].apply(extract_threads)
```

```python
def extract_core_breakdown(x):
    if pd.isna(x):
        return pd.Series(
            [np.nan, np.nan, np.nan],
            index=["p_cores", "e_cores", "lp_e_cores"]
        )
    
    s = str(x)

    # Find text inside parentheses: e.g. "(8P + 12E)" or "(6P + 8E + 2LP-E)"
    m = re.search(r"\(([^)]*)\)", s)
    if not m:
        return pd.Series(
            [np.nan, np.nan, np.nan],
            index=["p_cores", "e_cores", "lp_e_cores"]
        )

    inner = m.group(1)  # e.g. "8P + 12E", "6P + 8E + 2LP-E"

    # P-cores
    p_cores = np.nan
    m_p = re.search(r"(\d+)\s*P\b", inner, flags=re.IGNORECASE)
    if m_p:
        p_cores = int(m_p.group(1))

    # LP-E cores
    lp_e_cores = np.nan
    m_lp = re.search(r"(\d+)\s*LP-?E\b", inner, flags=re.IGNORECASE)
    if m_lp:
        lp_e_cores = int(m_lp.group(1))

    # E-cores: remove LP-E part first to avoid counting it as plain E
    inner_no_lp = re.sub(r"\d+\s*LP-?E\b", "", inner, flags=re.IGNORECASE)

    e_cores = np.nan
    m_e = re.search(r"(\d+)\s*E\b", inner_no_lp, flags=re.IGNORECASE)
    if m_e:
        e_cores = int(m_e.group(1))

    return pd.Series(
        [p_cores, e_cores, lp_e_cores],
        index=["p_cores", "e_cores", "lp_e_cores"]
    )

# Apply to your cores column
df1[["p_cores", "e_cores", "lp_e_cores"]] = df1["cores"].apply(extract_core_breakdown)
```

```python
import re
import pandas as pd

def parse_gpu(text: str):
    """
    Parse 'graphics' column into:
      - gpu_brand
      - gpu_series
      - gpu_model
      - gpu_vram  (int GB)
      - is_integrated (bool)
    """
    # Completely missing graphics info
    if pd.isna(text):
        return {
            "gpu_brand": None,
            "gpu_series": None,
            "gpu_model": None,
            "gpu_vram": None,
            "is_integrated": None,
        }

    s = str(text).strip()
    s_clean = re.sub(r"\s+", " ", s)
    s_lower = s_clean.lower()

    # ---------- VRAM ----------
    vram = None
    m_vram = re.search(r"(\d+)\s*gb", s_lower)
    if m_vram:
        vram = int(m_vram.group(1))

    # ---------- BRAND ----------
    brand = None
    if "nvidia" in s_lower:
        brand = "NVIDIA"
    elif "amd" in s_lower:
        brand = "AMD"
    elif "qualcomm" in s_lower or "adreno" in s_lower:
        brand = "Qualcomm"
    elif "intel" in s_lower:
        brand = "Intel"
    elif "iris xe" in s_lower:
        # sometimes only "Iris Xe Graphics" appears
        brand = "Intel"
    elif "mediatek" in s_lower:
        brand = "MediaTek"
    elif "mali-" in s_lower:
        brand = "ARM"
    elif "core gpu" in s_lower:
        # e.g. "10 Core GPU", "Apple 8 Core GPU"
        brand = "Apple"

    series = None
    model = None

    # strip VRAM prefix like "6 GB "
    body = re.sub(r"^\s*\d+\s*gb\s+", "", s_clean, flags=re.IGNORECASE)
    body_no_brand = body
    if brand:
        body_no_brand = re.sub(brand, "", body, flags=re.IGNORECASE).strip()
    b_low = body_no_brand.lower()

    # ---------- NVIDIA ----------
    if brand == "NVIDIA":
        if "geforce rtx" in b_low:
            series = "GeForce RTX"
        elif "geforce gtx" in b_low:
            series = "GeForce GTX"
        elif "geforce mx" in b_low:
            series = "GeForce MX"
        elif "geforce" in b_low:
            series = "GeForce"
        elif "rtx" in b_low:
            series = "RTX"

        # RTX/GTX/MX patterns
        m_model = (
            re.search(r"(rtx\s+[a-z]*\d+\w*)", b_low) or
            re.search(r"(gtx\s+[a-z]*\d+\w*)", b_low) or
            re.search(r"(mx\s+\d+\w*)", b_low)
        )
        if m_model:
            model = m_model.group(0).upper()
        else:
            m_fallback = re.search(r"(geforce\s+.+)", body_no_brand, flags=re.IGNORECASE)
            if m_fallback:
                model = m_fallback.group(0).strip()

    # ---------- AMD ----------
    elif brand == "AMD":
        m_rad = re.search(r"(radeon\s+.+)", body_no_brand, flags=re.IGNORECASE)
        if m_rad:
            rad_part = m_rad.group(1).strip()
            series = "Radeon"
            after = re.sub(r"radeon", "", rad_part, flags=re.IGNORECASE).strip()
            model = after if after else "Radeon"
        elif "radeon" in b_low:
            series = "Radeon"
            model = "Radeon"
        else:
            if "graphics" in b_low or "integrated" in b_low:
                series = "Graphics"
                model = "AMD Graphics"

    # ---------- INTEL ----------
    elif brand == "Intel":
        b_low = body_no_brand.lower()
        if "iris xe" in b_low:
            series = "Iris Xe"
            model = "Iris Xe Graphics"
        elif "uhd" in b_low:
            series = "UHD"
            m_uhd = re.search(r"uhd(?:\s+graphics)?\s*(\d+)?", b_low)
            if m_uhd:
                if m_uhd.group(1):
                    model = f"UHD {m_uhd.group(1)}"
                else:
                    model = "UHD Graphics"
        elif "hd graphics" in b_low or "intel hd" in b_low:
            series = "HD"
            model = "HD Graphics"
        elif "arc" in b_low:
            series = "Arc"
            m_arc = re.search(r"(arc\s+.+)", body_no_brand, flags=re.IGNORECASE)
            model = m_arc.group(1).strip() if m_arc else "Arc Graphics"
        else:
            if "graphics" in b_low or "integrated" in b_low or "hd" in b_low:
                series = "Integrated Graphics"
                model = "Intel Integrated Graphics"

    # ---------- Qualcomm / Adreno ----------
    elif brand == "Qualcomm":
        series = "Adreno"
        m_adr = re.search(r"(adreno\s+\w+)", b_low, flags=re.IGNORECASE)
        model = m_adr.group(1).title() if m_adr else "Adreno GPU"

    # ---------- MediaTek / ARM ----------
    elif brand in ("MediaTek", "ARM"):
        m_mali = re.search(r"(mali-\w+)", s_lower)
        if m_mali:
            series = "Mali"
            model = m_mali.group(0).upper()
        else:
            series = "Integrated Graphics"
            model = s_clean

    # ---------- Apple ----------
    elif brand == "Apple":
        series = "Apple GPU"
        model = s_clean

    # ---------- Fallback for unknown brand ----------
    else:
        if "integrated" in s_lower and "graphics" in s_lower:
            series = "Integrated Graphics"
            model = s_clean

    # ---------- is_integrated ----------
    is_integrated = None
    if "dedicated" in s_lower:
        is_integrated = False
    elif "integrated" in s_lower:
        is_integrated = True
    elif brand == "NVIDIA":
        # NVIDIA is almost always discrete in your dataset
        is_integrated = False
    elif brand in ("Intel", "AMD", "Qualcomm", "Apple", "MediaTek", "ARM"):
        # if no VRAM mentioned, treat as integrated; if VRAM mentioned, treat as discrete
        is_integrated = (vram is None)

    return {
        "gpu_brand": brand,
        "gpu_series": series,
        "gpu_model": model,
        "gpu_vram": vram,
        "is_integrated": is_integrated,
    }

# ---- Apply on your df1 ----
gpu_parsed = df1["graphics"].apply(parse_gpu).apply(pd.Series)
df1[["gpu_brand", "gpu_series", "gpu_model", "gpu_vram", "is_integrated"]] = \
    gpu_parsed[["gpu_brand", "gpu_series", "gpu_model", "gpu_vram", "is_integrated"]]

# If you want to verify:
# df1[df1["graphics"].notna() & df1["is_integrated"].isna()].shape
```

```python
def parse_model_smart(name: str):
    """
    Extract ONLY:
    - series_name
    - device_type
    with special handling for:
      * HP 15s-/14s-/15- style
      * HP 240R G9 / 250R G9
      * Chromebooks (Asus / Acer / Lenovo)
      * Asus Gaming V16 ...
      * Acer Aspire Lite ...
      * Simple code-only models (Walker NU15A1, Dell DC15255)
    """
    if pd.isna(name):
        return {
            "series_name": None,
            "device_type": None
        }

    s = str(name).strip()
    s_clean = re.sub(r"\s+", " ", s)
    s_lower = s_clean.lower()

    # Only use part before "(" (after that are specs)
    main_part = s_clean.split("(", 1)[0].strip()
    tokens = main_part.split()

    if not tokens:
        return {
            "series_name": None,
            "device_type": None
        }

    brand = tokens[0]

    # ---------------- DEVICE TYPE ----------------
    device_type = "General"

    gaming_keywords = ["gaming", "rog", "tuf", "legion", "omen", "victus",
                       "predator", "alienware", "zephyrus"]

    business_keywords = ["thinkpad", "latitude", "probook", "elitebook",
                         "thinkbook", "vostro", "240r g9", "250r g9"]

    convertible_keywords = ["2-in-1", "2 in 1", "convertible", "flip", "x360", "yoga"]

    thin_light_keywords = ["slim", "air", "swift", "vivobook", "go", "gram"]

    if any(k in s_lower for k in gaming_keywords):
        device_type = "Gaming"
    elif any(k in s_lower for k in convertible_keywords):
        device_type = "Convertible"
    elif any(k in s_lower for k in business_keywords):
        device_type = "Business"
    elif any(k in s_lower for k in thin_light_keywords):
        device_type = "Thin & Light"

    # ---------------- SERIES NAME ----------------
    STOP_WORDS = {"laptop", "notebook", "chromebook"}  # note: "gaming" handled via special case

    def is_year_token(t: str) -> bool:
        return bool(re.fullmatch(r"20\d\d", t))

    def is_heavy_code(t: str) -> bool:
        """Very long codey tokens we don't want in series (A9UK7PT##ACJ, etc.)."""
        t_clean = re.sub(r"[^A-Za-z0-9]", "", t)
        return len(t_clean) >= 7 and re.search(r"[A-Za-z]", t_clean) and re.search(r"\d", t_clean)

    series_tokens = []

    # ------------ Brand-specific rules ------------

    # 1) HP special patterns
    if brand == "HP" and len(tokens) >= 2:
        t1 = re.sub(r"[,:;()]+$", "", tokens[1])

        # a) HP 15-fd0468TU / 15s-fc0156AU / 14s-dq3141TU etc.
        m_hp_hyphen = re.match(r"^(\d+s?)\-", t1, re.IGNORECASE)
        if m_hp_hyphen:
            series_tokens = [m_hp_hyphen.group(1)]  # '15', '15s', '14s'
            return {
                "series_name": " ".join(series_tokens),
                "device_type": device_type
            }

        # b) HP 15 2025 → want "15" only (avoid year)
        if len(tokens) >= 3 and re.fullmatch(r"\d{2,3}", tokens[1]) and is_year_token(tokens[2]):
            series_tokens = [tokens[1]]
            return {
                "series_name": " ".join(series_tokens),
                "device_type": device_type
            }

        # c) HP 240R G9 A9UK7PT##ACJ Laptop → "240R G9"
        if len(tokens) >= 3 and re.match(r"^\d+[A-Za-z]$", tokens[1]) and re.match(r"^G\d+$", tokens[2], re.IGNORECASE):
            series_tokens = [tokens[1], tokens[2]]
            return {
                "series_name": " ".join(series_tokens),
                "device_type": device_type
            }

    # 2) Chromebooks: Asus/Acer/Lenovo Chromebook CX1500...
    if brand in ("Asus", "Acer", "Lenovo") and len(tokens) >= 2 and tokens[1].lower() == "chromebook":
        # e.g., "Asus Chromebook CX1500FKA-E80077 Laptop"
        if len(tokens) >= 3 and not is_heavy_code(tokens[2]) and tokens[2].lower() not in STOP_WORDS:
            series_tokens = [tokens[1], tokens[2]]
        else:
            series_tokens = [tokens[1]]
        return {
            "series_name": " ".join(series_tokens),
            "device_type": device_type
        }

    # 3) Asus Gaming V16 ...
    if brand == "Asus" and len(tokens) >= 3 and tokens[1].lower() == "gaming" and tokens[2].lower().startswith("v"):
        # "Asus Gaming V16 2025 V3607VH-RP038WS Laptop..."
        series_tokens = [tokens[1], tokens[2]]  # "Gaming V16"
        return {
            "series_name": " ".join(series_tokens),
            "device_type": device_type
        }

    # 4) Acer Aspire Lite ...
    #    "Acer AL15-52H Aspire Lite 15 Laptop..."
    if brand == "Acer":
        for i, t in enumerate(tokens):
            if t.lower() == "aspire":
                # Take from "Aspire" up to "Laptop" or other stop word
                end = len(tokens)
                for j in range(i, len(tokens)):
                    if tokens[j].lower() in STOP_WORDS:
                        end = j
                        break
                series_tokens = tokens[i:end]  # e.g. ["Aspire", "Lite", "15"]
                return {
                    "series_name": " ".join(series_tokens),
                    "device_type": device_type
                }

    # ------------ Generic fallback ------------

    # Collect tokens after brand until a STOP_WORD or heavy code
    for i, t in enumerate(tokens[1:], start=1):
        t_clean = re.sub(r"[,:;()]+$", "", t)
        t_low = t_clean.lower()

        if t_low in STOP_WORDS:
            break

        if is_heavy_code(t_clean):
            break

        series_tokens.append(t_clean)

    # If still nothing, fall back to just the second token (like DC15255, NU15A1)
    if not series_tokens and len(tokens) >= 2:
        series_tokens = [re.sub(r"[,:;()]+$", "", tokens[1])]

    series_name = " ".join(series_tokens) if series_tokens else None

    return {
        "series_name": series_name,
        "device_type": device_type
    }

# ---- APPLY TO df1 ----
model_parsed = df1["model"].apply(parse_model_smart).apply(pd.Series)

df1[["series_name", "device_type"]] = model_parsed[["series_name", "device_type"]]
```

```python
import re
import numpy as np

def clean_gpu(row):
    raw = str(row["graphics"]).strip()

    # Handle missing
    if raw.lower() in ["nan", "none", ""]:
        return pd.Series([None, None, None, None])

    text = raw.lower()

    # ===== NVIDIA =====
    if "rtx" in text or "nvidia" in text or "geforce" in text:
        brand = "NVIDIA"

        # Series
        if "rtx" in text:
            series = "GeForce RTX"
        elif "mx" in text:
            series = "GeForce MX"
        else:
            series = "GeForce"

        # GPU Model extraction
        model = re.findall(r"(rtx\s?\d+\w*|mx\s?\d+|rtx\s?a?\d+\w*)", text, flags=re.I)
        model = model[0].upper().replace(" ", "") if model else None

        # Remove memory part ("8GB", etc.)
        if model:
            model = model.replace("GB", "").strip()

        is_integrated = False

        return pd.Series([brand, series, model, is_integrated])

    # ===== AMD =====
    if "amd" in text or "radeon" in text:
        brand = "AMD"

        # Integrated AMD (e.g. Radeon 820M, Radeon Graphics)
        if "radeon" in text and not any(x in text for x in ["rx", "m", "xt", "pro", "vega"]):
            series = "Radeon"
            model = "Radeon Graphics"
            is_integrated = True
        else:
            # Discrete AMD GPU (RX / PRO)
            series = "Radeon"
            model_match = re.findall(r"(rx\s?\d+\w*)", text, flags=re.I)
            model = model_match[0].upper().replace(" ", "") if model_match else "Radeon"
            is_integrated = False

        return pd.Series([brand, series, model, is_integrated])

    # ===== INTEL =====
    if "intel" in text or "iris" in text or "uhd" in text or "hd graphics" in text or "integrated graphics" in text:

        brand = "Intel"

        # Series
        if "iris" in text:
            series = "Iris Xe"
            model = "Iris Xe Graphics"
        elif "uhd" in text:
            series = "UHD"
            model = "UHD Graphics"
        elif "hd graphics" in text:
            series = "HD"
            model = "HD Graphics"
        elif "arc" in text:
            series = "Arc"
            model = "Arc Graphics"
        else:
            # Generic "Intel Integrated" / "Intel Graphics"
            series = "Integrated Graphics"
            model = "Integrated Graphics"

        is_integrated = True

        return pd.Series([brand, series, model, is_integrated])

    # ===== Qualcomm Adreno =====
    if "adreno" in text:
        brand = "Qualcomm"
        series = "Adreno"
        model = "Adreno GPU"
        is_integrated = True
        return pd.Series([brand, series, model, is_integrated])

    # ===== Apple GPU =====
    if "apple" in text:
        brand = "Apple"
        series = "Apple GPU"

        # Extract model: "10-core GPU", "8-core GPU"
        model = re.findall(r"(\d+\-?core gpu)", text)
        model = model[0].title() if model else "Apple GPU"

        is_integrated = True
        return pd.Series([brand, series, model, is_integrated])

    # ===== Fallback =====
    return pd.Series([None, None, None, None])


# ==== APPLY CLEANING ====
df1[["gpu_brand", "gpu_series", "gpu_model", "is_integrated"]] = df1.apply(clean_gpu, axis=1)
```

```python
import re

def fix_gpu_model_spacing(model):
    if model is None or str(model).lower() == "none" or str(model).strip() == "":
        return model
    
    m = str(model)

    # === NVIDIA RTX ===
    # RTX4060 → RTX 4060
    m = re.sub(r"RTX\s?(\d+)", r"RTX \1", m, flags=re.I)

    # RTXA500 → RTX A500
    m = re.sub(r"RTX\s?A\s?(\d+)", r"RTX A\1", m, flags=re.I)
    m = re.sub(r"RTXA\s?(\d+)", r"RTX A\1", m, flags=re.I)

    # RTX3050A → RTX 3050A
    m = re.sub(r"RTX\s?(\d+)(A)", r"RTX \1\2", m, flags=re.I)

    # RTX5070TI → RTX 5070 Ti
    m = re.sub(r"RTX\s?(\d+)\s?TI", r"RTX \1 Ti", m, flags=re.I)

    # === AMD RX ===
    # RX6500M → RX 6500M
    m = re.sub(r"RX\s?(\d+\w*)", r"RX \1", m, flags=re.I)

    # === NVIDIA MX ===
    # MX230 → MX 230
    m = re.sub(r"MX\s?(\d+)", r"MX \1", m, flags=re.I)

    # === NVIDIA RTX (clean double spaces) ===
    m = re.sub(r"\s+", " ", m).strip()

    return m

df1["gpu_model"] = df1["gpu_model"].apply(fix_gpu_model_spacing)
```

```python
import re

def refine_gpu_row(row):
    brand   = row["gpu_brand"]
    series  = row["gpu_series"]
    model   = row["gpu_model"]
    text    = str(row["graphics"]) if "graphics" in row else ""

    # Default: keep current values
    new_series = series
    new_model  = model

    # ---------- AMD: recover RX models like RX 6500M ----------
    if brand == "AMD":
        m_rx = re.search(r"(RX\s*\d+\w*)", text, flags=re.IGNORECASE)
        if m_rx:
            # e.g. "RX 6500M" -> "RX 6500M"
            new_model = m_rx.group(1).upper().replace("  ", " ")
            # optionally mark series as Radeon RX (if you want more granularity)
            # new_series = "Radeon RX"
            return new_series, new_model

    # ---------- Intel Arc: keep 130V / 140V model codes ----------
    if brand == "Intel" and series == "Arc":
        # e.g. "Arc 130V GPU (8GB)", "Arc 140T GPU"
        m_arc = re.search(r"Arc\s+(\d+\w*)", text, flags=re.IGNORECASE)
        if m_arc:
            code = m_arc.group(1).upper()
            new_model = f"Arc {code} GPU"
            return new_series, new_model

    # ---------- Qualcomm Adreno ----------
    if brand == "Qualcomm":
        # Normalize everything to "Adreno GPU"
        return "Adreno", "Adreno GPU"

    # ---------- Intel integrated generic ----------
    if brand == "Intel" and series == "Integrated Graphics":
        # Clean up model like "Intel Integrated" -> "Integrated Graphics"
        return new_series, "Integrated Graphics"

    # ---------- Apple GPU names ----------
    if brand == "Apple" and isinstance(model, str):
        # Normalize "10-Core GPU" -> "10 Core GPU", keep "Apple " prefix
        # ensure we always have "Core" with a space, not "Core-"
        m_core = re.search(r"(\d+)\s*-?\s*Core", model, flags=re.IGNORECASE)
        if m_core:
            cores = m_core.group(1)
            new_model = f"Apple {cores} Core GPU"
        else:
            # fallback: keep as is but replace "-Core" with " Core"
            new_model = model.replace("-Core", " Core")
        return new_series, new_model

    # Default: return existing values
    return new_series, new_model

# ---- Apply refinement in-place ----
df1["gpu_series"], df1["gpu_model"] = zip(*df1.apply(refine_gpu_row, axis=1))
```
